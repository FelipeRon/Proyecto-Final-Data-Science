# Evidencia del replay de producción
Ejecuta el runner según el README para regenerar el replay directamente desde Snowflake. Este notebook revisa las exportaciones generadas por esa ejecución.

```python
from pathlib import Path
import json
import pandas as pd
docs = Path("docs") if Path("docs").is_dir() else Path("../docs")
proof = json.loads((docs / "verificacion_rubrica.json").read_text(encoding="utf-8"))
batches = pd.read_csv(docs / "monitoreo.csv")
assert len(batches) == proof["snowflake"]["periods"] == 8
assert batches.n.sum() == proof["snowflake"]["replay_predictions"] == 2000
print("Query ID:", proof["query_id"])
print("Lotes:", len(batches), "Predicciones:", batches.n.sum())
print("Alerta de deriva:", batches.loc[batches.deriva, "periodo"].tolist())
```
